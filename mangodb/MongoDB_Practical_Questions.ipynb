# MongoDB Practical Questions - Jupyter Notebook


```python
# 1. Load Superstore CSV into MongoDB
import pandas as pd
from pymongo import MongoClient
client = MongoClient('mongodb://localhost:27017/')
db = client['superstore_db']
orders = db['Orders']
df = pd.read_csv('Superstore.csv')
orders.insert_many(df.to_dict('records'))
print('Data inserted successfully')
```

```python
# 2. Retrieve and print all documents
for doc in orders.find():
    print(doc)
```

```python
# 3. Count total documents
print(orders.count_documents({}))
```

```python
# 4. Fetch orders from West region
for doc in orders.find({'Region':'West'}):
    print(doc)
```

```python
# 5. Orders where Sales > 500
for doc in orders.find({'Sales':{'$gt':500}}):
    print(doc)
```

```python
# 6. Top 3 orders with highest Profit
for doc in orders.find().sort('Profit',-1).limit(3):
    print(doc)
```

```python
# 7. Update Ship Mode
result = orders.update_many({'Ship Mode':'First Class'},{'$set':{'Ship Mode':'Premium Class'}})
print(result.modified_count)
```

```python
# 8. Delete orders where Sales < 50
result = orders.delete_many({'Sales':{'$lt':50}})
print(result.deleted_count)
```

```python
# 9. Total sales per region
pipeline=[{'$group':{'_id':'$Region','TotalSales':{'$sum':'$Sales'}}}]
for doc in orders.aggregate(pipeline):
    print(doc)
```

```python
# 10. Distinct Ship Modes
print(orders.distinct('Ship Mode'))
```

```python
# 11. Count orders for each category
pipeline=[{'$group':{'_id':'$Category','OrderCount':{'$sum':1}}}]
for doc in orders.aggregate(pipeline):
    print(doc)
```
